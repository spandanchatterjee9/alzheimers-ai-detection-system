# 🧠 Multimodal Alzheimer's AI: Vision + Clinical Text with Dynamic Gating

### Architecture:
1. **Vision Branch:** Multi-Kernel ResNet-18 (`focal_multikernel_resnet_best.pt` - 76.00% Baseline) -> Flattening Layer -> 512-dim 1D vector ($W_I$)
2. **Text Branch:** Bio_ClinicalBERT trained on structured patient clinical reports -> 768-dim 1D vector ($W_T$)
3. **Top Decision Model:** Dynamic Gating Network learning patient-specific priority weights ($h = g \odot v + (1-g) \odot t$) to resolve borderline MCI ambiguity.
